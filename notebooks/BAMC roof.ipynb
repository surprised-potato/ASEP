```python
from anastruct import SystemElements

# 1. Initialize
ss = SystemElements()

# 2. Define Geometry
# Element 1: (0,0) to (4, 1.4)
el1_id = ss.add_element(location=[[0, 0], [4, 1.4]])
# Element 2: (4, 1.4) to (6.8, 2.4)
el2_id = ss.add_element(location=[[4, 1.4], [6.8, 2.4]])

# 3. Add Supports
ss.add_support_hinged(node_id=1)
ss.add_support_roll(node_id=2, direction=2) # Vertical roller

# 4. Apply Loads
# Gravity load (downward y-axis)
ss.q_load(q=-2300, element_id=el1_id, direction='y')
ss.q_load(q=-2300, element_id=el2_id, direction='y')

# 5. Solve
ss.solve()

# 6. Display Results
print("Structure Setup:")
# Use show_structure() instead of show_reaction_force() to avoid the crash
ss.show_structure() 

print("Bending Moment:")
ss.show_bending_moment()

print("Shear Force:")
ss.show_shear_force()

print("Displacement:")
ss.show_displacement()

# Add this to the end of your script to see the reaction numbers
for node_id in ss.reaction_forces:
    print(f"Node {node_id} Reaction Force: {ss.reaction_forces[node_id]}")
```
